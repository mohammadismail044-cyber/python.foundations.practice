 Variables and data types

```python
name = "Ismail"
age = 20
height = 1.75
is_student = True
print(name, age, height, is_student)
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

 Operators

```python
a = 17
b = 5
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
print("Is a greater than b?", a > b)
print("Both conditions:", a > 10 and b < 10
```

Conditional statements

```python
marks = int(input("Enter marks (0-100): "))
if marks >= 80:
print("Grade: A")
elif marks >= 70:
print("Grade: B")
elif marks >= 60:
print("Grade: C")
else:
print("Needs improvement"
```

Loops

```python
# for loop
for number in range(1, 6):
print(number)
# while loop
count = 1
while count <= 5:
print("Count:", count)
count += 1
```

Strings

```python
message = "Python Programming"
print(message.upper())
print(message.lower())
print(message.title())
print("Length:", len(message))
print("First character:", message[0])
print("First word:", message[0:6])
print("Python" in message
```

Lists and tuples

```python

subjects = ["Python", "Math", "English"]
subjects.append("Computer Science")
subjects[1] = "Statistics"
print(subjects)
print("Number of subjects:", len(subjects))
coordinates = (10, 20)
print("Tuple:", coordinates)
```

Dictionaries and sets

```python
student = {"name": "Ismail", "age": 20, "department": "CS"}
print(student["name"])
student["semester"] = 3
print(student)
unique_numbers = {1, 2, 2, 3, 3, 4}
print(unique_numbers)
```

Functions

```python
def greet(name):
return "Hello, " + name
def calculate_area(length, width):
return length * width
print(greet("Ismail"))
print("Area:", calculate_area(5, 4))
```

 Error handling

```python
try:
number = float(input("Enter a number: "))
print("Double:", number * 2)
except ValueError:
print("Please enter a valid number.")
```
